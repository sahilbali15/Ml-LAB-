# ***Standardization in machine learning is a preprocessing technique that rescales numerical features so they have a mean of zero and a standard deviation of one***

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/Social_Network_Ads (1).csv")
```

```python
df
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df=df.iloc[:,2:]
```

```python
from sklearn.model_selection import train_test_split
```

```python
df.columns
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.drop('Purchased',axis=1),df['Purchased'],test_size=0.3,random_state=0)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
sc.fit(x_train)
```

```python
x_train_scaled=sc.transform(x_train)
x_test_scaled=sc.transform(x_test)
```

```python
sc.mean_
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# before scaling
ax1.set_title('Before Scaling')
sns.kdeplot(x_train['Age'], ax=ax1)
sns.kdeplot(x_train['EstimatedSalary'], ax=ax1)

# after scaling
ax2.set_title('After Standard Scaling')
sns.kdeplot(x_train_scaled[:, 0], ax=ax2) # Access 'Age' column by index
sns.kdeplot(x_train_scaled[:, 1], ax=ax2) # Access 'EstimatedSalary' column by index
plt.show()
```
